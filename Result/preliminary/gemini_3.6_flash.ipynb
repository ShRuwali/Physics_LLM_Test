```python
import os
import base64
from pathlib import Path

import requests
from dotenv import load_dotenv
```

```python
load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")
if not api_key:
    raise ValueError("OPENROUTER_API_KEY is missing from your .env file.")
```

```python
# Replace with the path to your local MP4 video
video_path = Path("SlidingBlock.mp4")
```

```python
video_base64 = base64.b64encode(video_path.read_bytes()).decode("utf-8") ## encode the video bytes as text for JSON request
video_data_url = f"data:video/mp4;base64,{video_base64}"
```

```python
response = requests.post(
    url="https://openrouter.ai/api/v1/chat/completions",
    headers={
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json",
    },
    json={
        "model": "google/gemini-2.5-flash",
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": (
                            "Describe what happens in this video in 2 sentence. "
                            
                        ),
                    },
                    {
                        "type": "video_url",
                        "video_url": {"url": video_data_url},
                    },
                ],
            }
        ],
    },
    timeout=180,
)
```

```python
result = response.json()
print(result["choices"][0]["message"]["content"])
```

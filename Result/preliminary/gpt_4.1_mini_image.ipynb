```python
import os
import sys
from dotenv import load_dotenv
from openai import OpenAI
import base64
from IPython.display import Image
```

```python
openai= OpenAI()

api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found.")
elif not api_key.startswith("sk-proj-"):
    print("There is an API key but it does not start with sk-proj-")
elif api_key.strip() != api_key:
    print("There is an api key but there may be some space or tab characters at the start or end ")
else:
    print("API key found.")
```

```python
PATH=os.path.join('..','..','Problems','Image','Problem_1_Regular.png')

Image(PATH, width=400)
```

```python
image_path = PATH

# Function to encode the image
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

```

```python
# Getting the Base64 string
base64_image = encode_image(image_path)
```

```python
response = openai.responses.create(
    model="gpt-4.1-mini",
    input=[
        {
            "role": "user",
            "content": [
                {"type": "input_text", "text": "what's in this image?"},
                {
                    "type": "input_image",
                    "image_url": f"data:image/png;base64,{base64_image}",
                },
            ],
        }
    ],
)

print(response.output_text)
```
